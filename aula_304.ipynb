#Corrigir dados alterando o tipo dos dados

.apply(sum) -> posso aplicar uma função em cada linha

lambda -> cria uma função

x.strip('USD' ) -> remove caracteres do começo

#Não esquecer que para salvar deve declarar como VAR


```python
df['usd_goal_real'] = df['usd_goal_real'].apply(lambda x: x.strip('USD '))
```

```python
df.head()
```

Remoção de um valor com um padrão utilizando o regex

```python
df['launched'].str.replace(' \d\d:\d\d','', regex=True)
```

```python
df['launched'] = df['launched'].str.replace(' \d\d:\d\d','', regex=True)
```

```python
df.head()
```

Transformando o tipo dos dados

.astype('int')
int - inteiro
float - numero flutuante
string - texto
bool - booleando (v e f)

```python
df['usd_goal_real'] = df['usd_goal_real'].astype('int64')
```

para declarar data usamos o format=' ' , é dizer essa string é uma data, transforme ela em uma data

para data com pandas -> busca no google pandas datetime format codes

```python
df['deadline'] = pd.to_datetime(df['deadline'], format='%d/%m/%y')
df['launched'] = pd.to_datetime(df['launched'], format='%d/%m/%y')
```

nesse processo vamos agrepar por 1 o que for de um determinado e 0 se não for de um determinado estado  
get_dumminies

codigo com o processo de conversao

```python
df_invest = pd.get_dummies(df_invest, columns=['backedLocation'])
df_invest = df_invest.astype(int)
```

```python
df_invest.head()
```

```python
df_invest.sort_values (by=['ID'])
```

feito media para ter uma estimativa sobre media de estafo e idade que ajudaram

```python
df_invest = df_invest.groupby(by=['ID']).agg({'age':'mean','backedLocation_BR':'sum', 'backedLocation_GBK':'sum','backedLocation_US':'sum'})
```

```python
df_invest.head()
```

```python
df = df.merge(df_invest, how='right', on='ID')
```
